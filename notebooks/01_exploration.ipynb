7 : Lazy evaluation

La lazy evaluation c'est le fait que Spark ne fait aucun calcul immédiatement quand on lui demande de modifier des données.

À la place, il enregistre simplement la liste des instructions à effectuer. Il n'exécute ces instructions sur les données que lorsqu'on lui demande un résultat final. 

Cela permet à Spark d'optimiser les étapes pour aller plus vite et économiser de la mémoire.

Différence entre Transformation et action:
Une Transformation prépare ou modifie les données sans lancer de calcul (ex: filtrer, trier). Elle crée simplement une nouvelle action.
Une Action demande un résultat précis ou une sauvegarde. C'est elle qui déclenche réellement tous les calculs

Exemples:

3 exemples de Transformations (pas de calcul immédiat) :
1. `filter()` : garder seulement certaines lignes.
2. `select()` : choisir certaines colonnes.
3. `drop()` : surimer une colonne.

3 exemples d'Actions (déclenchent les calculs) :
1. `count()` : compter le nombre de lignes.
2. `show()` : afficher les premières lignes à l'écran.
3. `write` : sauvegarder les données dans un fichier (ex: CSV ou Parquet).ex: CSV ou Parquet).
